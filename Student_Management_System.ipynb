```python
import pandas as pd
#  Create a dataframe
students={
    'Name':['John', 'Mary','Peter','Grace','David','Sarah','James','Linda'],
    'Math':[85,90,65,95,72,88,55,80],
    'English':[78,88,70,92,75,90,60,85],
    'Science':[92,85,60,96,70,84,58,82],
    'Attendance':[95,98,80,99,85,97,75,90]
}
df=pd.DataFrame(students)
df.to_csv('Results.csv')
df
```

```python
import pandas as pd
# Load dataset into pandas
df=pd.read_csv('Results.csv')
df

# DATA CLEANING
# Exploring the first five columns
print('================================================================')
ED=df.head()
print(ED)
# Exploring the last five columns
print('================================================================')
ED1=df.tail()
print(ED1)
# Exploring dataset information
print('================================================================')
ED2=df.info()
print(ED2)
# Exploring dataset description
print('================================================================')
ED3=df.describe()
print(ED3)
# Identifying dataset shape
print('================================================================')
ED4=df.shape 
print(ED4)
# Trimming the dataset
print('================================================================')
df=df.drop(columns='Unnamed: 0')
df

# Checking the number of students in the dataset
print('================================================================')
ED5=df['Name'].count()
print(f'Number of students in the dataset: {ED5}')

# Calculating the average math score
print('================================================================')
ED6=df['Math'].mean()
print(f'Average Math Score: {ED6}')

# Finding out who scored the highest in science
print('================================================================')
ED7=df['Science'].max()
ED8=df[df['Science']==ED7]
print(f'Highest Score in Science: {ED7}')
print(f'Student Who score the Highest in Science: \n {ED8}')

# Students with attendance below 85%
print('================================================================')
ED9=df[df['Attendance']<85]
print(f'Student with Attendance below 85%: \n {ED9}')

# =============== FEATURE ENGINEERING ==================================
# Adding Total Marks column
print('================================================================')
df['Total Marks']=df['Math']+df['English']+df['Science']
df
# Adding and calculating the average column
df['Average']=df['Total Marks']/3
df
df['Average']=df['Average'].round(2)
df
# Sorting students from the largest
df=df.sort_values(by=['Total Marks'], ascending=False)
df

# Grading the students
# Creating a Grade function
def grade(score):
    if score>=80:
        return 'A'
    elif score>=70:
        return 'B'
    elif score>=60:
        return 'C'
    elif score>=50:
        return 'D'
    else:
        return 'F'
    
# Applying the funtion to the dataset
df['Grade']=df['Average'].apply(grade)
df
```
