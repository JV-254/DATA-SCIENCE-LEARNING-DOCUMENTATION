```python
import pandas as pd
import numpy as np
import statistics

scores=[65,72,80,55,90,68,75,82,60,70]

# Create a dataframe
df=pd.DataFrame({
    'Scores':scores
})
print(df)
print(df.columns)
```

```python
# Calculating the mean
mean=df["Scores"].mean()
print("Mean: ",mean)
```

```python
# Calculating Median
median=df['Scores'].median()
print(f'Median: ',median)
```

```python
# Calculating mode
mode=df['Scores'].mode()
print(f'Mode: ', mode)
```

```python
# Calculating Range
range=df['Scores'].max()- df['Scores'].min()
print(f'Range: ',range)
```

```python
# Calculating the variance
variance=df['Scores'].var()
print(f'Variance: ',variance)
```

```python
# Standard Deviation
Standard_Deviation=df['Scores'].std()
print(f'Standard Deviation: ',Standard_Deviation)
```

```python
# Minimum and Maximum
minimum=df['Scores'].min()
maximum=df['Scores'].max()
print(f'Minimum: ',minimum)
print(f'Maximum: ',maximum)
```

```python
# Quartiles
Q1=df['Scores'].quantile(0.25)
Q2=df['Scores'].quantile(0.50)
Q3=df['Scores'].quantile(0.75)
print(f'Q1: ',Q1)
print(f'Q2: ',Q2)
print(f'Q3: ',Q3)
```

```python
# Interquatile Range
IQR=df['Scores'].quantile(0.75)-df['Scores'].quantile(0.25)
print(f'Interquatile Range: ',IQR)
IQR1=Q3-Q1
print(IQR1)
```

```python
# Calculating everything at once
description=df['Scores'].describe()
print(description)
```
