```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
# Loading dataframe
df=pd.read_csv('Students.csv')
df

# Data Analysis and manipulation
# Feature Engineering
# Adding total Marks column
df['Total Marks']=df['Math']+df['English']+df['Science']
df
# Adding Average Score Column
df['Average Score']=(df['Total Marks']/3).round(2)
df

# Grading the students
def grade(score):
    if score>=80:
        return 'A'
    elif score>=70:
        return 'B'
    elif score>=60:
        return 'C'
    elif score>=50:
        return 'D'
    else:
        return 'E'
    
# Applying the function to the dataframe
df['Grade']=df['Average Score'].apply(grade)
df

# Create a bar chart of average marks by subject
subjects=['Math','English','Science']
average_Marks=[
    df['Math'].mean(),
    df['English'].mean(),
    df['Science'].mean()
]
# Creating a font type and size
font1={
    'family':'Arial',
    'color':'darkred',
    'size':20
}
font2={
    'family':'serif',
    'color':'blue',
    'size':15
}
# Plot the graph
plt.figure(figsize=(6,4))
plt.bar(subjects,average_Marks, color='brown')
plt.title('Average Marks per student', fontdict=font1)
plt.xlabel('Subjects', fontdict=font2)
plt.ylabel('Average Score', fontdict=font2)
plt.show()
# A function to determine if a student passed or failed
def status(score1):
    if score1>=85:
        return 'Pass'
    else: 
        return 'Fail'
    
# Applying the function to the dataframe
df['Status']=df['Average Score'].apply(status)
df

# PI chart for student who failed or passed
status_count=df['Status'].value_counts()
plt.pie(status_count, labels=status_count.index, autopct='%1.1f%%')
plt.title('PIE CHART')
plt.show()
```
