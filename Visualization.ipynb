```python
# Types of visualizations
# 1, Line Chart
import matplotlib.pyplot as plt
# Creating the x and y coordinates
x=[10,20,30,40]
y=[20,25,35,55]

# Plotting the values
plt.plot(x,y, marker='o',c='Brown', mfc='b', ls='dashed',lw='10.5')
# Creating a title
plt.title('LINE CHART')
plt.ylabel('Y-Axis')
plt.xlabel('X-Axis')
plt.show()
```

```python
# Bar Chart
import matplotlib.pyplot as plt
# Coordinates
x=['Thur','Fri','Sat','Sun']
y=[170,120,250,190]

# Plotting the values
plt.barh(x,y,color='r',height=0.3)
plt.title('BAR CHART')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

```python
# Histogram
import matplotlib.pyplot as plt
# Coordinates
x=[7,8,9,10,10,12,12,12,13,14,14,15,16,16,17,18,18,19,20,20,21,22,23,24,25,25,26,28,30,32,35,35,36,38,40,42,44,48,50]
# Plot the values
# Creating font types
font1={
    'family':'Arial',
    'color':'blue',
    'size':20
}

font2={
    'family':'serif',
    'color':'darkred',
    'size':15
}
plt.hist(x,bins=10, color='Blue')
plt.title('HISTOGRAM', fontdict=font1, loc='left')
plt.xlabel('Total Bill', fontdict=font2)
plt.ylabel('Frequency', fontdict=font2)
plt.show()
```

```python
# Scatter Plot
import matplotlib.pyplot as plt
# Coordinates
x=[10,15,20,25,30]
y=[12,18,25,28,35]

a=[13,17,23,29,35]
b=[11,15,22,27,32]
# Creating font types
font1={
    'family':'Arial',
    'color':'blue',
    'size':20
}

font2={
    'family':'serif',
    'color':'darkred',
    'size':15
}
# Plot the gragh
plt.scatter(x,y,c='r')
plt.scatter(a,b)
plt.title('SCATTER PLOT CHART', fontdict=font1, loc='left')
plt.xlabel('Study Hours',fontdict=font2)
plt.ylabel('Marks',fontdict=font2)
plt.show()
```

```python
# PIE CHART
import matplotlib.pyplot as plt
# create data
cars=[
    'AUDI','BMW','FORD','TESLA','JAGUAR'
]
 
data=[
    23,10,35,15,12
]

# Draw the pie chart
myexplode=[0.2,0.3,0.1,0,0]
plt.pie(data, labels=cars, autopct='%1.1f%%', explode=myexplode)
# Slicing the pie chart
plt.title('PIE CHART')
plt.show()
```

```python
# BOX PLOT
import matplotlib.pyplot as plt
import numpy as np
# Create an array
data=np.array([
    [10,12,14,15,18,20,22],
    [8,9,11,13,17,19,21],
    [14,16,18,20,23,25,27]
])
# Creating a font type and size
font1={
    'family':'Arial',
    'color':'brown',
    'size':20,
    'weight':'bold'
}
font2={
    'family':'serif',
    'color':'blue',
    'size':15
}
# Box plot
plt.boxplot(data)
plt.title('BOX PLOT', fontdict=font1, loc='left')
plt.xlabel('Groups',fontdict=font2)
plt.ylabel('Values', fontdict=font2)
plt.show()
```

```python
# HEATMAP
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(0)
data=np.random.rand(10,10)
plt.imshow(data,cmap='viridis',interpolation='nearest')

plt.colorbar()
plt.xlabel('X-axis Label')
plt.ylabel('Y-axis Label')
plt.title('HEATMAP')
plt.show()
```

```python
# 3D LINE PLOT
from mpl_toolkits import mplot3d
import numpy as np
import matplotlib.pyplot as plt

fig = plt.figure()
ax = plt.axes(projection='3d')

z = np.linspace(0, 100, 101)
x = z * np.sin(25 * z)
y = z * np.cos(25 * z)

ax.plot3D(x, y, z, 'green')
ax.set_title('3D Line Plot')
plt.show()
```
