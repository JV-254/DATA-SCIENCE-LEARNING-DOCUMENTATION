```python
# Writting a dataframe to csv format
import pandas as pd
# Create a dataframe
data={
    'Name':['Aparna','Pankaj','Sudhir','Geeku'],
    'Degree':['MBA','BCA','M.Tech','MBA'],
    'Score':[90,40,98,70]
}
# Creating a dataframe
df=pd.DataFrame(data)
# Creating a csv file format
df.to_csv('file1.csv')
print(df)
# Removing the index column
df.to_csv('file2.csv',index=False)
# Exporting only the selected columns
df.to_csv('file3.csv',columns=['Name'], index=False)
# Excluding headers
df.to_csv('file4.csv',index=False,header=False)
# Handling missing values
df.to_csv('file5.csv',index=False,na_rep='nothing')
# Changing the separater
df.to_csv('file6.csv',index=False, sep='\t')
```
