```python
import pandas as pd
# Creating a data Frame
data={
    'Name':['John', 'Mary','Peter','Grace','David','Sarah','James','Linda'],
    'Math':[85,90,65,95,72,88,55,80],
    'English':[78,88,70,92,75,90,60,85],
    'Science':[92,85,60,96,70,84,58,82],
    'Attendance':[95,98,80,99,85,97,75,90]
}
df=pd.DataFrame(data)
print(df)
df.to_csv('Students.csv', index=False)
```

```python
import pandas as pd
# Loading the dataset
df=pd.read_csv('Students.csv')
print(df)
# Eploring the dataset 
# First five rows
print('---------------------------------------------------------------------------------------------------')
ED=df.head()
print(ED)
# Last five rows
print('---------------------------------------------------------------------------------------------------')
ED1=df.tail()
print(ED1)
# Checking the dataset information
print('---------------------------------------------------------------------------------------------------')
ED2=df.info()
print(ED2)
# Getting statistics summarry
print('---------------------------------------------------------------------------------------------------')
ED3=df.describe()
print(ED3)
# Checking the shape of the dataframe
print('---------------------------------------------------------------------------------------------------')
ED4=df.shape
print(ED4)
# Viewing colunm names
print('---------------------------------------------------------------------------------------------------')
ED5=df.columns
print(ED5)
# How many students are in the dataset
print('---------------------------------------------------------------------------------------------------')
ED6=df['Name'].count()
print(f'Number of students: {ED6}')
# Average Math score
print('---------------------------------------------------------------------------------------------------')
ED7=df['Math'].mean()
print(f"Average Math Score:{ED7}")
# Who scored the highest in science
ED8=df['Science'].max()
print(f'Highest score in Science: {ED8}')
# Student with attendance below 85%
ED9=df[df['Attendance']<85]
print(f'Student with Attendance less than 85: \n {ED9}')
```
