```python
# STUDENT PERFORMANCE ANALYZER
import pandas as pd
# Creating a DataFarme
data={
    'Name':['John', 'Mary','Peter','Grace','David','Sarah','James','Linda'],
    'Math':[85,90,65,95,72,88,55,80],
    'English':[78,88,70,92,75,90,60,85],
    'Science':[92,85,60,96,70,84,58,82],
    'Attendance':[95,98,80,99,85,97,75,90]
}
df=pd.DataFrame(data)
df
# Reading the first 5 rows
ED=df.head()
print(ED)
# Reading the last 5 rows of the dataFrame
print('-------------------------------------------------------------------------------------------------')
ED1=df.tail()
print(ED1)
# Viewing the information of the dataset
print('-------------------------------------------------------------------------------------------------')
ED3=df.info()
print(ED3)
# Viewing the description of the dataset
print('-------------------------------------------------------------------------------------------------')
ED4=df.describe()
print(ED4)
# Viewing the shape of the dataFrame
print('-------------------------------------------------------------------------------------------------')
ED5=df.shape
print(ED5)
# Viewing the number of columns in the dataFrame
print('-------------------------------------------------------------------------------------------------')
ED6=df.columns
print(ED6)

# CHECKING FOR NULL VALUES
print('-------------------------------------------------------------------------------------------------')
ED7=df.isnull()
print(ED7)
print('-------------------------------------------------------------------------------------------------')
ED8=df.isna()
print(ED8)

# DATA ANALYSIS AND MANIPULATION
# How many Students are in the dataset
print('-------------------------------------------------------------------------------------------------')
ED9=df['Name'].count()
print(f'Number of Students: {ED9}')

# Average Math Score
print('-------------------------------------------------------------------------------------------------')
ED10=df['Math'].mean()
print(f'Average Math Score: {ED10}')

# Student who scored the highest in science
print('-------------------------------------------------------------------------------------------------')
ED11=df['Science'].max()
ED12=df[df['Science']==ED11]
print(f"Student who scored highest in Science: \n {ED12}")

# Student with attendance below 85
print('-------------------------------------------------------------------------------------------------')
ED13=df[df['Attendance']<85]
print(ED13)

# Feature Engineering 
# Adding other columns
# Total Marks column
df['Total Marks']=df['Math']+df['English']+df['Science']
df
# Creating Average Column
df['Average Score']=df['Total Marks']/3
df['Average Score']=df['Average Score'].round(2)

#   Finding student with the highest total a score
print('-------------------------------------------------------------------------------------------------')
ED14=df['Total Marks'].max()
ED15=df[df['Total Marks']==ED14]
print(ED15)

# Ranking students from the highest to the lowest
print('-------------------------------------------------------------------------------------------------')
ED16=df.sort_values(by=['Total Marks'], ascending=False)
print(ED16)

# Grading students
# Create a grade function
def grade(Score):
    if Score>=80:
        return 'A'
    elif Score>=70:
        return 'B'
    elif Score>=60:
        return 'C'
    elif Score>=50:
        return 'D'
    else:
        return 'F'
    
# Calling the function
df['Grade']=df['Average Score'].apply(grade)
df

# Converting the dataFrame to csv
df.to_csv('Marks.csv')
df

# Identifying the correlation
```
