```python
import numpy as np
import pandas as pd
# Create a 1-D array
arr=np.array([2,4,5,6,7,8,9])
# Create a pandas series
pd_Series=pd.Series(arr,index=("a","b","c","d","e","f","g"))
print(pd_Series)
# Accessing elements in Series data
x=pd_Series["c"]
print(x)

# Creating a dataframe
data={
    "Calories":[480,350,690,654],
    "Duration":[12,43,67,87]
}
df=pd.DataFrame(data, index=("a","b","c","d"))
print(df)

# Accessing one dataframe
access=df.loc[["b","c","d"]]
print(access)

# Creating a series from a range
ser=pd.Series(range(5,20))
print(ser)

# Accessing elements using slicing
y=ser[2:5]
print(y)

# Binary operations in a series
# Arithmetic operations
s1=pd.Series([10,20,30,40,50], index=('a','b','c','d','e'))
s2=pd.Series([1,20,3,40,5], index=('a','b','c','d','e'))
# Adding the two series
results=s1+s2
print(results)
# Comparision operations
results1=s1==s2
print(results1)

# Binary operations on pandas dataframe
data1={
    "A":[34,65,76,87,46],
    "B":[23,76,12,43,28]
}
data2={
    "C":[23,65,86,00,78],
    "D":[12,65,98,34,18]
}
df1=pd.DataFrame(data1, index=('a','b','c','d','e'))
df2=pd.DataFrame(data2, index=('a','b','c','d','e'))
print(df1)
print(df2)

# Adding the dataframes
results3=pd.DataFrame({
    "A+C":df1['A']+df2['C'],
    "B+D":df1['B']+df2['D']
})
print(results3)
```
