```python
# PERSONAL EXPENSE TRACKER
import pandas as pd
# Creating a data Frame
Expense={
    'Date':['2025-06-01','2025-06-01','2025-06-02','2025-06-03','2025-06-04','2025-06-05','2025-06-06','2025-06-07','2025-06-08','2025-06-09'],
    'Category':['Food','Transport','Food','Entertainment','Utilities','Food','Transport','Shopping','Food','Entertainment'],
    'Description':['Lunch','Bus Fare','Breakfast','Movie','Internet','Dinner','Taxi','T-shirt','Lunch','Games'],
    'Amount':[350,100,200,800,2500,600,500,1200,400,700],
    'Payment Method':['Cash','Mpesa','Cash','Mpesa','Mpesa','Cash','Mpesa','Card','Cash','Mpesa']
}
df=pd.DataFrame(Expense)
df

# Having an overview of the first 5 rows
ED1=df.head()
print(ED1)
print('===========================================================================================')
# Viewing the last 5 crows
ED2=df.tail()
print(ED2)
# Viewing the information of the dataset
print('===========================================================================================')
ED3=df.info()
print(ED3)
# Viewing the description of the dataset
print('===========================================================================================')
ED4=df.describe()
print(ED4)
# Identifying the shape of the Dataset
print('===========================================================================================')
ED5=df.shape
print(ED5)
# Identifying the columns in the dataset
print('===========================================================================================')
ED6=df.columns
print(ED6)
# Number of Expenses
print('===========================================================================================')
ED7=df.shape[0]
print(f'Number of Expenses: {ED7}')

# Total Amount of money spent
print('===========================================================================================')
ED8=df['Amount'].sum()
print(f'Total Amount of Money Spent: {ED8}')

# Calculating the average Expense
print('===========================================================================================')
ED9=df['Amount'].mean()
print(f'Average Expense: {ED9}')

# Finding the largest Expense
print('===========================================================================================')
ED10=df['Amount'].max()
print(f'Largest Expense: {ED10}')

# Smallest Expense
print('===========================================================================================')
ED11=df['Amount'].min()
print(f'Smallest Expense: {ED11}')

# CATEGORY ANALYSIS
# Total Spending per category
print('===========================================================================================')
ED12=df.groupby('Category')['Amount'].sum()
print(f'Total Spending Per Category: \n {ED12}')

# Highest Spending category
print('===========================================================================================')
ED13=df.groupby('Category')['Amount'].sum().idxmax()
print(f'Highest Spending Per Category: \n {ED13}')

# Filtering Data
# All Food Expensense
print('===========================================================================================')
ED14=df[df['Category']=='Food']
print(f'All Food Expenses: \n {ED14}')

# Expenses Above 1000
print('===========================================================================================')
ED15=df[df['Amount']>1000]
print(f'Expenses Above 1000: \n {ED15}')


# CREATING A MONTHLY SUMMARY
# Converting the Date column
print('===========================================================================================')
df['Date']=pd.to_datetime(df['Date'])
df

# Extracting the month
print('===========================================================================================')
df['Month']=df['Date'].dt.month
df

# Extracting Month Name
df['Month Name']=df['Date'].dt.month_name()
# Extracting letters from the date
df['Month Name']=df['Date'].dt.strftime('%b')
# Extracting the year
df['Year']=df['Date'].dt.year
# Changing the date format
df['Date1']=df['Date'].dt.strftime('%d/%m/%Y')
# Total Spending per month
print('===========================================================================================')
ED16=df.groupby('Month')['Amount'].sum()
print(f'Total Spending Per Month: \n {ED16}')
df
print('===========================================================================================')
ED17=df.info()
print(ED17)
df

# Expenses per Category
print('===========================================================================================')
ED18=df.groupby('Category')['Amount'].count()
print(f'Expenses Per Category: \n {ED18}')

# Top 3 most expensive transactions
print('===========================================================================================')
ED19=df.nlargest(3,'Amount')
print(f'Top 3 largest Transactions: \n {ED19}')

# Percentage for spending for each category
print('===========================================================================================')
ED20=df['Amount'].sum()
ED21=(df.groupby('Category')['Amount'].sum()/ED20)*100
print(ED21)

# All Expenses paid using M-pesa
print('===========================================================================================')
ED22=df[df['Payment Method']=='Mpesa']
print(f'All Expenses paid using M-pesa: \n {ED22}')

# Expense Level
# Create a function
def Expense(level):
    if level<500:
        return 'Low'
    elif level<1000:
        return 'Medium'
    else:
        return 'High'
# Create a column called Expense Level
df['Expense Level']=df['Amount'].apply(Expense)
df

# Sorting expenses from the highest to the lowest
print('===========================================================================================')
ED22=df.sort_values(by='Amount',ascending=False)
print(ED22)
df
    
```
