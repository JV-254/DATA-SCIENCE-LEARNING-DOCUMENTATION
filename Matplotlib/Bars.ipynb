```python
# Plotting multiple bars
import matplotlib.pyplot as plt
import numpy as np

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales_2025 = [12000, 15000, 13000, 18000, 22000, 25000]
sales_2026 = [14000, 16000, 15000, 21000, 24000, 28000]

x=np.arange(len(months))
width=0.35

fig,ax=plt.subplots(figsize=(12,6))

bars_2025=ax.bar(x-width/2,
                 sales_2025,
                 width,
                 color='blue',
                 label='2025 sales')
bars_2026=ax.bar(x+width/2,
                 sales_2026,
                 width,
                 color='red',
                 label='2026 sales')
ax.set_title('2025 Sales VS 2026 Sales')
ax.set_xlabel('Month')
ax.set_ylabel('Sale (Ksh)')
ax.set_xticks(x)

ax.set_xticklabels(months)
ax.bar_label(bars_2025)
ax.bar_label(bars_2026)

ax.grid(axis='y')
ax.legend()

plt.tight_layout()
plt.show()
```

```python
# Sorting data for visualization
import matplotlib.pyplot as plt

courses = ["Computer Science", "Data Science", "Software Engineering", "Networking"]
students = [45, 30, 50, 25]

# Combining both list
data=list(zip(courses,students))
# Sorting the data
data=sorted(data, key=lambda x:x[1], reverse=True)
# Separating sorted data 
sorted_courses=[item[0] for item in data]
sorted_students=[item[1] for item in data]

fig,ax=plt.subplots(figsize=(12,6))

bars=ax.bar(sorted_courses,
       sorted_students,
       color='red')
ax.set_title('Sorted Courses and Students')
ax.set_xlabel('Course')
ax.set_ylabel('Number of Students')

ax.bar_label(bars)
ax.grid(axis='y')
plt.tight_layout()
plt.show()


```

```python
# Exercise
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales = [12000, 15000, 13000, 18000, 22000, 25000]

# Combining both lists
data=list(zip(months,sales))
# sorting them
data=sorted(data, key=lambda x:x[1], reverse=True)

sorted_months=[item[0] for item in data]
sorted_sales=[item[1] for item in data]

fig, ax=plt.subplots(figsize=(12,6))

bars=ax.bar(sorted_months,
       sorted_sales,
       color='blue')
ax.set_title('Total Sales Per month')
ax.set_xlabel('Month')
ax.set_ylabel('Number of sales')

ax.bar_label(bars)
ax.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
# Stacked Bar chart
import matplotlib.pyplot as plt

courses = ["Computer Science", "Data Science", "Software Engineering", "Networking"]

male = [25, 18, 30, 15]
female = [20, 12, 20, 10]

fig, ax=plt.subplots(figsize=(12,6))

ax.bar(courses,male, label='Male',color='red')
ax.bar(courses,female, bottom=male, label='Female',color='blue')

ax.set_title('Students By Course and Gender')
ax.set_xlabel('Course')
ax.set_ylabel('Number of Students')

ax.grid(axis='y')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Excersice
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

online_sales = [5000, 7000, 6000, 9000, 12000, 15000]

store_sales = [7000, 8000, 7000, 9000, 10000, 10000]

fig,ax=plt.subplots(figsize=(12,6))

online=ax.bar(months,
       online_sales,
       label='Online',
       color='brown')
store=ax.bar(months,
       store_sales, 
       bottom=online_sales,
       label='Store',
       color='green')

ax.set_title('Monthly Sales Channel')
ax.set_xlabel('Month')
ax.set_ylabel('Sales (Ksh)')
ax.bar_label(online)
ax.bar_label(store)

ax.legend()
ax.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
# 100% Stacked bar chart
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

online_sales = [5000, 7000, 6000, 9000, 12000, 15000]

store_sales = [7000, 8000, 7000, 9000, 10000, 10000]

# Calculating the total
total_sales=[
    online+store
    for online,store in zip(online_sales,store_sales)
]

# Calculating the online percentage
online_percent=[
    round(online/total*100,2)
    for online,total in zip(online_sales,total_sales)
]

# Calculating store percentage
store_percent=[
    round(store/total*100,2)
    for store,total in zip(store_sales,total_sales)
]

fig,ax=plt.subplots(figsize=(12,6))

online=ax.bar(months,
       online_percent,
       color='blue',
       label='online')
store=ax.bar(months,
             store_percent,
             bottom=online_percent,
             color='red',
             label='store')

ax.set_title('Monthly Sales Composition')
ax.set_xlabel('Months')
ax.set_ylabel('Percentage (%)')
ax.set_ylim(0,100)

ax.legend()
ax.bar_label(online)
ax.bar_label(store)

ax.grid(axis='y')

plt.tight_layout()
plt.show()
```
