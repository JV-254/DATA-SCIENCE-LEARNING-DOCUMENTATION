```python
# Plotting multiple bars
import matplotlib.pyplot as plt
import numpy as np

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales_2025 = [12000, 15000, 13000, 18000, 22000, 25000]
sales_2026 = [14000, 16000, 15000, 21000, 24000, 28000]

x=np.arange(len(months))
width=0.35

fig,ax=plt.subplots(figsize=(12,6))

bars_2025=ax.bar(x-width/2,
                 sales_2025,
                 width,
                 color='blue',
                 label='2025 sales')
bars_2026=ax.bar(x+width/2,
                 sales_2026,
                 width,
                 color='red',
                 label='2026 sales')
ax.set_title('2025 Sales VS 2026 Sales')
ax.set_xlabel('Month')
ax.set_ylabel('Sale (Ksh)')
ax.set_xticks(x)

ax.set_xticklabels(months)
ax.bar_label(bars_2025)
ax.bar_label(bars_2026)

ax.grid(axis='y')
ax.legend()

plt.tight_layout()
plt.show()
```
