```python
# Plotting histograms
import matplotlib.pyplot as plt

scores = [45, 52, 55, 60, 62, 65, 68, 70, 72, 75,
          78, 80, 82, 85, 88, 90, 92, 95]

fig, ax=plt.subplots(figsize=(12,6))

ax.hist(scores)

ax.set_title('Distribution Of student scores')
ax.set_xlabel('Score')
ax.set_ylabel('Number of Students')

ax.grid(axis='y')
plt.tight_layout()
plt.show()
```
