```python
import matplotlib.pyplot as plt
import numpy as np

scores=[
    45,52,55,60,62,65,68,70,72,75,78,80,82,85,88,90,92,95
]
scores.append(150)
fig,ax=plt.subplots(figsize=(12,6))

ax.boxplot(scores,
           patch_artist=True,
           showmeans=True,
           meanline=False,
           notch=False,
           vert=False,
           boxprops=dict(facecolor='skyblue',color='black'),
           medianprops=dict(color='r',linewidth=2),
           whiskerprops=dict(color='black',linewidth=1.5),
           capprops=dict(color='black'),
           meanprops=dict(marker='o',markerfacecolor='g'))

ax.set_title('Distribution of student scores')
ax.set_ylabel('Score')
ax.grid(axis='x')

# Calculating the percentile ranges
percentile_25th=np.percentile(scores,25)
percentile_50th=np.percentile(scores,50)
percentile_75th=np.percentile(scores,75)

# Calculating IQR
IQR=percentile_75th-percentile_25th

print(f'Lower Percentile: ',percentile_25th)
print(f'Median: ',percentile_50th)
print(f'Upper Percentile: ',percentile_75th)
print(f'Inter Quartile Range: ',IQR)

# Finding an outlier
lower_bound=percentile_25th-(1.5*IQR)
upper_bound=percentile_75th+(1.5*IQR)

outlier=[
    score for score in scores
    if score<lower_bound or score>upper_bound
]
print(f'Lower Bound: ',lower_bound)
print(f'Upper Bound: ',upper_bound)
print(f'Outlier: ',outlier)
plt.tight_layout()
plt.show()
```

```python
# Creating pie charts
import matplotlib.pyplot as plt

channels=['Online','Physical Score', 'Wholesale']
sales=[50000,30000,20000]

fig,ax=plt.subplots(figsize=(12,6))
# Showing both values and percentages
def show_value_and_percent(percent):
    total=sum(sales)
    value=percent*total/100
    return f'{value: ,.0f}\n{percent: .1f}%'

ax.pie(
    sales,
    labels=channels,
    autopct=show_value_and_percent,
    startangle=90,
    explode=(0.1,0,0),
    shadow=False,
    colors=['skyblue','lightgreen','orange']
)
ax.set_title('Sales Distribution Channel')
ax.axis('equal')
plt.tight_layout()
plt.show()
```

```python
# Creating 4 subplots
import matplotlib.pyplot as plt 
import numpy as np

students = ['A', 'B', 'C', 'D', 'E']
scores = [65, 75, 80, 55, 90]
study_hours = [2, 4, 5, 1, 7]

fig,ax=plt.subplots(2,2, figsize=(12,6))

# first visual line chart between student and scores 
ax[0,0].plot(students,
             scores,
             marker='o',
             color='r')
ax[0,0].set_title('Students Vs Scores')
ax[0,0].set_xlabel('Score')
ax[0,0].set_ylabel('Scores')

# Second visual -bar chart between students and scores
ax[0,1].bar(students,
            scores,
            color='b')
ax[0,1].set_title('Students VS Score Bar chart')
ax[0,1].set_xlabel('Student')
ax[0,1].set_ylabel('Scores')

# Third visual - Scatter plot between scores and study hours
ax[1,0].scatter(study_hours,
                scores,
                color='green',
                alpha=0.7)
ax[1,0].set_title('Study Hours VS Score')
ax[1,0].set_xlabel('Study Hours')
ax[1,0].set_ylabel('Score')

# Drawing a trend line in between the scatter plot
# Calculating the trend line points
slope, intercept=np.polyfit(study_hours,scores,1)
trend=slope *np.array(study_hours) + intercept

# plotting the trend line
ax[1,0].plot(study_hours,trend, label='Trend Line',ls='-')

# Fourth visual - histogram
ax[1,1].hist(scores,bins=5,color='blue')
ax[1,1].set_title('Scores Distribution')
ax[1,1].set_xlabel('Score')
ax[1,1].set_ylabel('Number of Students')

# Identifying the type of distribution in the histogram
mean=np.mean(scores)
median=np.median(scores)

print(f'Mean: ',mean)
print(f'Median: ',median)

# Draw the lines
ax[1,1].axvline(mean,ls=':',c='g',lw=5,label='Mean Score')
ax[1,1].axvline(median,ls='-',c='r',lw='3',label='Median Score')

ax[1,0].legend()
ax[1,1].legend()
ax[1,1].legend()
plt.tight_layout()
plt.show()
```

```python
# Comparing multiple bars
import matplotlib.pyplot as plt
import numpy as np
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales_2025 = [12000, 15000, 13000, 18000, 22000, 25000]
sales_2026 = [14000, 16000, 15000, 21000, 24000, 28000]

x=np.arange(len(months))
width=0.35

fig,ax=plt.subplots(figsize=(12,6))

bars_2025=ax.bar(x-width/2,
                 sales_2025,
                 width,
                 color='blue',
                 label='Sales 2025')
bars_2026=ax.bar(x+width/2,
                 sales_2026,
                 width,
                 color='r',
                 label='Sales 2026')

ax.set_title('2025 VS 2026 SALES COMPARISION')
ax.set_xlabel('Month')
ax.set_ylabel('Sales (Ksh)')

ax.bar_label(bars_2025)
ax.bar_label(bars_2026)

ax.grid(axis='y')

plt.tight_layout()
plt.show()

```

```python
# Excersice
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [12000, 15000, 13000, 18000, 22000, 25000]
expenses = [8000, 9000, 8500, 11000, 14000, 16000]
profit = [4000, 6000, 4500, 7000, 8000, 9000]

fig,ax=plt.subplots(1,3, figsize=(12,6))

# First plot line chart
ax[0].plot(months,
             sales,
             c='r',
             lw=3)
ax[0].set_title('Monthly sales Trend')
ax[0].set_xlabel('Month')
ax[0].set_ylabel('Sales (Ksh)')

# Second plot bar chart
bars=ax[1].bar(months,
            expenses,
            color='blue')
ax[1].set_title('Monthly Expenses')
ax[1].set_xlabel('Month')
ax[1].set_ylabel('Expenses')

# Third plot line chart
ax[2].plot(months,
             profit,
             marker='o',
             ls='-',
             c='green',
             lw=4)
ax[2].set_title('Monthly Profit')
ax[2].set_xlabel('Month')
ax[2].set_ylabel('Profit (Ksh)')

ax[1].bar_label(bars)

fig.suptitle('Sales Dashboard')
plt.tight_layout()
plt.show()
```

```python
# Calculating category sales
# Create a dataframe
import pandas as pd
import matplotlib.pyplot as plt

data={
    'Product':['Laptop','Phone','Desk','Chair','Tablet'],
    'Category':['Electronics','Electronics','Furniture','Furniture','Electronics'],
    'Sales':[80000,45000,25000,15000,30000]
}
# Create a dataframe
df=pd.DataFrame(data)
df 

# Grouping the data
category_sales=df.groupby('Category')['Sales'].sum()
print(category_sales)

# Visualizing
fig, ax=plt.subplots(figsize=(12,6))
category_sales.plot(
    kind='bar',
    ax=ax,
    color='r'
)

ax.set_title('Total Sales in each category')
ax.set_xlabel('Category')
ax.set_ylabel('Total  Sales(Ksh)')
ax.grid(axis='y')
plt.tight_layout()
plt.show()
```
