```python
# ADDING A LEGEND
import matplotlib.pyplot as plt
import numpy as np

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [12000, 15000, 13000, 18000, 22000, 25000]

plt.plot(months,sales,
         c='g',
         ls='--',
         lw=2,
         marker='o',
         ms=7,
         label='Monthly Sales'
         )
plt.title('Monthly Sales Performance')
plt.xlabel('Month')
plt.ylabel('Sales(Ksh)')
plt.grid()
plt.legend()
plt.show()
```

```python
# PLOTTING MULTIPLE LINES
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales_2025 = [12000, 15000, 13000, 18000, 22000, 25000]

sales_2026 = [14000, 16000, 15000, 21000, 24000, 28000]

plt.plot(months, sales_2025, label='2025 Sales',c='b',ls='--',marker='o')
plt.plot(months,sales_2026, label='2026 Sales',c='g',ls='-',marker='s')
plt.title('MONTHLY SALES')
plt.xlabel('Month')
plt.ylabel('Sales(Ksh)')
plt.grid()
plt.legend()
plt.show()
```

```python
# Excercise
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales_2025 = [12000, 15000, 13000, 18000, 22000, 25000]

sales_2026 = [14000, 16000, 15000, 21000, 24000, 28000]
plt.figure(figsize=(12,6))
plt.plot(months,sales_2025,marker='o', label='2025 Sales',c='b',ls='--')
plt.plot(months,sales_2026,marker='s', label='2026 Sales',c='g',ls='-')

# Annotation
plt.annotate('Highest Sale', 
             xy=('June',28000),
             xytext=('May',25000),
             arrowprops=dict(arrowstyle='->')
             )
plt.ylim(0,30000)
plt.xticks(rotation=45)
plt.title('2025 vs 2026 Monthly sales')
plt.xlabel('Month')
plt.ylabel('Sales(Ksh)')

plt.grid()
plt.legend()
plt.savefig('Sales_Comparision.png', 
            dpi=300,
            bbox_inches='tight')
plt.show()
```
