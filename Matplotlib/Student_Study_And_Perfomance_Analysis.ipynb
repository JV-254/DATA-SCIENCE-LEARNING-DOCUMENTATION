```python
#STUDENT STUDY AND PERFOMANCE ANALYSIS
import matplotlib.pyplot as plt

students = ["Ann", "Brian", "Carol", "David", "Eric",
            "Faith", "George", "Hellen", "Ian", "Jane"]

study_hours = [2, 3, 4, 5, 6, 7, 8, 4, 6, 9]

scores = [50, 55, 62, 68, 72, 78, 85, 60, 75, 92]

# LINE PLOT
fig, ax=plt.subplots(figsize=(12,6))

ax.plot(students,scores,
        marker='o')

ax.set_title('Student Scores')
ax.set_xlabel('Student')
ax.set_ylabel('Scores')
ax.tick_params(axis='x',rotation=45)

# Finding student with the highest score
highest_score=max(scores)
highest_index=scores.index(highest_score)
highest_student=students[highest_index]
ax.annotate(f'Highest Score: {highest_student} ({highest_score})',
    xy=(highest_student,highest_score),
    xytext=('Ian',75),
        arrowprops=dict(arrowstyle='->')
    )
# Finding the student with the lowest score
lowest_score=min(scores)
lowest_index=scores.index(lowest_score)
lowest_student=students[lowest_index]
ax.annotate(f'Lowest Score: {lowest_student} ({lowest_score})',
            xy=(lowest_student,lowest_score),
            xytext=('Brian',55),
            arrowprops=dict(arrowstyle='->'))
ax.grid()
plt.tight_layout()
plt.show()
```

```python
#PART B- SCATTER PLOT
import matplotlib.pyplot as plt
import numpy as np
students = ["Ann", "Brian", "Carol", "David", "Eric",
            "Faith", "George", "Hellen", "Ian", "Jane"]

study_hours = [2, 3, 4, 5, 6, 7, 8, 4, 6, 9]

scores = [50, 55, 62, 68, 72, 78, 85, 60, 75, 92]

# Scatter plot showing study hours  vs score
fig, ax=plt.subplots(figsize=(12,6))

ax.scatter(study_hours,
           scores,
           s=100,
           alpha=0.7)

ax.set_title('Study Hours vs Scores')
ax.set_xlabel('Study Hours')
ax.set_ylabel('Scores')
 # Calculating correlation
correlation=np.corrcoef(study_hours,scores)[0,1]
print(f'Correlation: ',correlation)
ax.grid()
plt.show()
```

```python
# CREATING TREND LINES
import matplotlib.pyplot as plt 
import numpy as np

students = ["Ann", "Brian", "Carol", "David", "Eric",
            "Faith", "George", "Hellen", "Ian", "Jane"]

study_hours = [2, 3, 4, 5, 6, 7, 8, 4, 6, 9]

scores = [50, 55, 62, 68, 72, 78, 85, 60, 75, 92]

fig,ax=plt.subplots(figsize=(12,6))

ax.scatter(study_hours,
           scores,
           s=100,
           alpha=0.7,
           label='Student')

ax.set_title('Study Hours VS Scores')
ax.set_xlabel('Study Hours')
ax.set_ylabel('Scores')
# Drawing a trend line
slope,intercept=np.polyfit(study_hours,scores,1)
trend=slope * np.array(study_hours) + intercept

# Trend line plot
ax.plot(
    study_hours,
    trend,
    ls='-',
    label='Trend Line'
)
ax.legend()
ax.grid()
plt.tight_layout()
plt.show()
```

```python
# Drawing a trend line
import numpy as np
import matplotlib.pyplot as plt

study_hours = [2, 4, 3, 5, 6, 1, 7, 4]

scores = [55, 70, 65, 80, 90, 45, 95, 75]

fig, ax=plt.subplots(figsize=(12,6))

# Scatter plot
ax.scatter(study_hours,
           scores,
           s=100,
           alpha=0.7,
           label='Student')

# Calculate the trend line values
slope,intercept=np.polyfit(study_hours,scores,1)

trend=slope * np.array(study_hours) + intercept

# Trend line plot
ax.plot(study_hours,
        trend,
        label='Trend')

ax.set_title('Study Hours VS Scores')
ax.set_xlabel('study_hours')
ax.set_ylabel('Scores')

ax.grid()
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Calculating correlation and creating a scateer plot
import numpy as np
import matplotlib.pyplot as plt

study_hours = [2, 4, 3, 5, 6, 1, 7, 4]
scores = [55, 70, 65, 80, 90, 45, 95, 75]

correlation=np.corrcoef(study_hours,scores)[0,1]
print(f'Correlation: {correlation:.2f}')

fig,ax=plt.subplots(figsize=(12,6))

ax.scatter(study_hours,
           scores,
           s=100,
           alpha=0.7)

ax.set_title('Correlation between Study Hours and Scores')
ax.set_xlabel('Study Hours')
ax.set_ylabel('Scores')

ax.grid()
plt.tight_layout()
plt.show()
```

```python
# Bar charts
import matplotlib.pyplot as plt

products = ["Laptop", "Phone", "Tablet", "Monitor"]
sales = [25, 40, 20, 30]

fig,ax=plt.subplots(figsize=(12,6))

ax.bar(products,sales)

ax.set_title('Product Sales')
ax.set_xlabel('Product')
ax.set_ylabel('Sales')
ax.tick_params(rotation=45)

ax.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
# Excercise
import matplotlib.pyplot as plt

courses = ["Computer Science", "Data Science", "Software Engineering", "Networking"]
students = [45, 30, 50, 25]

fig,ax=plt.subplots(figsize=(12,6))

bars=ax.barh(courses,
       students,
       color='skyblue',
       edgecolor='black')
ax.set_title('Students By Course')
ax.set_xlabel('Course')
ax.set_ylabel('Number Of Students')
ax.bar_label(bars)
ax.grid(axis='x')
plt.tight_layout()
plt.show()
```

```python
# Plotting bar grapghs
import matplotlib.pyplot as plt

courses = ["Computer Science", "Data Science", "Software Engineering", "Networking"]
students = [45, 30, 50, 25]

fig,ax=plt.subplots(figsize=(12,6))

ax.bar(courses,
       students,
       color='blue',
      width=0.7)

ax.set_title('Number of students Per Course')
ax.set_xlabel('Course')
ax.set_ylabel('Students')
ax.grid(axis='y')

plt.tight_layout()
plt.show()
```

```python
# Plotting multiple bars side by side
import matplotlib.pyplot as plt
import numpy as np

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales_2025 = [12000, 15000, 13000, 18000, 22000, 25000]
sales_2026 = [14000, 16000, 15000, 21000, 24000, 28000]

x = np.arange(len(months))
width = 0.35

fig,ax=plt.subplots(figsize=(12,6))
bars_2025=ax.bar(
    x-width/2,
    sales_2025,
    width,
    label='2025 Sales'
)
bars_2026=ax.bar(
    x+width/2,
    sales_2026,
    width,
    label='2026 Sales'
)

ax.set_title('2025 VS 2026 Monthly Sales')
ax.set_xlabel('Month')
ax.set_ylabel('Sales')
ax.set_xticks(x)
ax.set_xticklabels(months)

ax.bar_label(bars_2025)
ax.bar_label(bars_2026)

ax.grid(axis='y')
ax.legend()

plt.tight_layout()
plt.show()
```
