```python
# Creating a dataframe
import pandas as pd
# Create a dataframe
myDataSet={
    'Cars':['Audi','Benz','Toyota','RR'],
    'Passings':[1,2,3,4]
}
myVar=pd.DataFrame(myDataSet)
print(myVar)

# Checking pandas version
print(pd.__version__)
print('--------------------------------------------------------')
# Creating a series
a=[10,20,30,40,50,60]
# create a dataseries
series=pd.Series(a,index=['a','b','c','d','e','f'])
print(series)
print('--------------------------------------------------------')
# Accessing element in data series
print(series['b'])
```
