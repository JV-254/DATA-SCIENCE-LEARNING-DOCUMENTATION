```python
# Net Salary Calculator
# 1. Payroll Number
PN=int(input("Enter Payroll Number: "))
# 2. Name
name=str(input("Enter Employee Name: "))
# 3. Gender
gender=str(input("Enter Employee Gender: "))
# 4. Derpartment
dept=str(input("Enter Employee Deparment: "))
# 5. Basic Salary
BasicSal=float(input("Enter Employee Basic Salary: "))
print(f'Payroll Number: ',PN)
print(f'Employee Name: ',name)
print(f'Employee Gender: ',gender)
print(f'Employee Department: ',dept)
print(f'Basic Salary: ',BasicSal)

# ALLOWANCES
houseAll=6500
medicalAll=5500
print('--------------------------------------------------------------------------')
# Calculating Gross Salary
Gross_Pay=BasicSal+houseAll+medicalAll
print(f'Gross Pay: ', Gross_Pay)

# Deductions
NHIF=(0.02*Gross_Pay)
NSSF=(0.03*Gross_Pay)
#Calculating P.A.Y.E
if(Gross_Pay>49000):
    PAYEE=(0.06*Gross_Pay)
    print(f'PAYEE: ',PAYEE)
elif(Gross_Pay>=31000 and Gross_Pay<=39000):
    PAYEE=(0.05*Gross_Pay)
    print(F'PAYEE: ',PAYEE)
elif(Gross_Pay>=20000 and Gross_Pay<=30000):
    PAYEE=(0.04*Gross_Pay)
    print(F'PAYEE: ',PAYEE)
elif(Gross_Pay>=0 and Gross_Pay<=19000):
    PAYEE=0
    print(F'PAYEE: ',PAYEE)
else:
    print('Error')
Total_Deductions=NHIF + NSSF + PAYEE
print(f'PAYEE: ',PAYEE)
print('--------------------------------------------------------------------------')
print(f'NHIF',NHIF)
print(f'NSSF',NSSF)
print(f'Total Deductions: ',Total_Deductions)

# Calculating Net Salary
Net_Pay=Gross_Pay - Total_Deductions
print('--------------------------------------------------------------------------')
print(f'Net Salary: ',Net_Pay)
print('--------------------------------------------------------------------------')
```
