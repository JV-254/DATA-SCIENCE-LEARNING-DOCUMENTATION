```python
import pandas as pd
# Creating a DataFrame
data={
    'Name':['John', 'Mary','Peter','Grace','David','Sarah','James','Linda'],
    'Math':[85,90,65,95,72,88,55,80],
    'English':[78,88,70,92,75,90,60,85],
    'Science':[92,85,60,96,70,84,58,82],
    'Attendance':[95,98,80,99,85,97,75,90]
}
df=pd.DataFrame(data)
print(df.to_string())
# Overview of the dataframe
# Viewing the first 5 rows
print('-----------------------------------------------------------------------------------------------------')
ED=df.head()
print(ED)
# Viewing the last 5 rows
print('-----------------------------------------------------------------------------------------------------')
ED1=df.tail()
print(ED1)
# Viewing information of the dataset
print('-----------------------------------------------------------------------------------------------------')
ED2=df.tail()
print(ED2)
# Viewing the dataset information
print('-----------------------------------------------------------------------------------------------------')
ED3=df.info()
print(ED3)
# Viewing description of the dataset
print('-----------------------------------------------------------------------------------------------------')
ED4=df.describe()
print(ED4)
# Getting the shape of the dataset
print('-----------------------------------------------------------------------------------------------------')
ED5=df.shape
print(ED5)
# Getting the columns of the Columns of the dataset
print('-----------------------------------------------------------------------------------------------------')
ED6=df.columns
print(ED6)
# Number of students in the dataset
print('-----------------------------------------------------------------------------------------------------')
ED7=df['Name'].count()
print(f'Number of Students in the dataFrame: {ED7}')
# Average Math score
print('-----------------------------------------------------------------------------------------------------')
ED8=df['Math'].mean()
print(f'Average Math Score: {ED8}')
# Highest Score in Science
print('-----------------------------------------------------------------------------------------------------')
ED9=df['Science'].max()
print(f'Highest Score in Science: {ED9}')
# Students whose attendance is below 85%
print('-----------------------------------------------------------------------------------------------------')
ED10=df[df['Attendance']<85]
print(f"Student Whose Attendance is below 85: \n {ED10}")

# Adding a new column Total Marks
df['Total Marks']=df['Math']+ df['English']+df['Science']
print('-----------------------------------------------------------------------------------------------------')
df
# Average Score for each student
df['Average Score']=df['Total Marks']/3
print('-----------------------------------------------------------------------------------------------------')
df
# Student with the highest total score
print('-----------------------------------------------------------------------------------------------------')
ED11=df['Total Marks'].max()
ED12=df[df['Total Marks']==ED11]
print(f'Student With the highest Total Score: \n {ED12}')
df
# Sorting Total Marks from the smallest'
print('-----------------------------------------------------------------------------------------------------')
ED13=df.sort_values(by='Total Marks', ascending=False)
print(ED13)

# Grading the students
def grade(Score):
    if Score>=80:
        return 'A'
    elif Score>=70:
        return 'B'
    elif Score>=60:
        return 'C'
    elif Score>=50:
        return 'D'
    else:
        return 'F'
# Creating a grade column
df['Grade']=df['Average Score'].apply(grade)
print(df[['Name','Average Score','Grade']])
```
